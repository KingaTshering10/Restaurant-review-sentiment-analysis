# Bag-of-Words, N-grams, and TF-IDF

Small illustrative examples of binary presence features, count features with unigrams through trigrams, and TF-IDF weights. The example sentences are toy documents, not restaurant-classifier training data. See [README.md](README.md) to continue with the classification notebook.


```python
import pandas as pd
import numpy as np
```

```python
doc1 = "Harry Potter is an amazing movie!!"
doc2 = "Harry Potter is the best movie!"
doc3 = "Harry potter is so great"
```

```python
# Using CountVectorizer, we get BOW features
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(binary=True) # Sentiment classification
doc1 = "Harry Potter is an amazing movie, harry!!"
doc2 = "Harry Potter is is the best movie!"
doc3 = "Harry potter is so great"
cv_out = cv.fit_transform([doc1,doc2,doc3])
```

```python
cv_out.toarray()
```

```python
pd.DataFrame(cv_out.toarray(), columns= cv.get_feature_names_out())
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(ngram_range=(1,3)) # unigram, bigram, trigram
# using ngrams we can get the context also. However, we see a large vector length
doc1 = "Harry Potter is an amazing movie , harry!!"
doc2 = "Harry Potter is the best movie!"
doc3 = "Harry potter is so great"
cv_out = cv.fit_transform([doc1,doc2,doc3])
pd.DataFrame(cv_out.toarray(), columns= cv.get_feature_names_out())
```

## TF-IDF

```python
from sklearn.feature_extraction.text import TfidfVectorizer
first_sent = "Data Science is an Amazing career in the current world"
second_sent = "Deep learning is a subset of machine learning"

vec = TfidfVectorizer()
result = vec.fit_transform([first_sent,second_sent])
result.toarray()
```

```python
pd.DataFrame(result.toarray(),columns= vec.get_feature_names_out())
```
